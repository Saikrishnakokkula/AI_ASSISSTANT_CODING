```python
def analyze_string(text):
    vowels = "aeiouAEIOU"
    vowel_count = 0
    consonant_count = 0
    digit_count = 0

    for char in text:
        if char in vowels:
            vowel_count += 1
        elif char.isalpha():
            consonant_count += 1
        elif char.isdigit():
            digit_count += 1

    return vowel_count, consonant_count, digit_count
```

```python
text = "Hello World 123"
result = analyze_string(text)
print("Vowels:", result[0])
print("Consonants:", result[1])
print("Digits:", result[2])
```

```python
def is_palindrome(text: str) -> bool:
    """
    Checks if a string is a palindrome, ignoring case and special characters.
    """
    # 1. Normalization: Remove non-alphanumeric and convert to lowercase
    # This ensures "Race Car" is recognized as a palindrome
    clean_text = "".join(char.lower() for char in text if char.isalnum())

    # 2. Logic: Compare the string to its inverse
    # Slicing [::-1] is the most efficient way to reverse in Python
    return clean_text == clean_text[::-1]

# Example usage with test cases
test_cases = ["racecar", "hello"]
for test in test_cases:
    print(f"'{test}': {is_palindrome(test)}")
```
